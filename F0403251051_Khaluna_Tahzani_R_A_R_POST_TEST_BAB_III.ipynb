```python
def hitung_diskon(harga, persen_diskon=10):
    potongan = harga * (persen_diskon / 100)
    harga_akhir = harga - potongan
    return harga_akhir
print(hitung_diskon(100000))
print(hitung_diskon(harga=100000, persen_diskon=25))
```

```python
def kurangi_stok(stok_saat_ini, jumlah_terjual):
  if jumlah_terjual > stok_saat_ini:
    return stok_saat_ini
  sisa = stok_saat_ini - jumlah_terjual
  return sisa
stok_gudang = 500
print(f"Stok awal sistem: {stok_gudang} unit")
stok_gudang = kurangi_stok(stok_gudang, 120)
print(f"Sisa stok di gudang setelah mutasi: {stok_gudang} unit")
```

```python
def input_data_panen():
# Simulasi data hasil panen dari beberapa petak lahan
    return [120, 95, 150, 80, 110]
def hitung_total_panen(data_panen):
    return sum(data_panen)
def hitung_rata_rata_panen(data_panen):
    return hitung_total_panen(data_panen) / len(data_panen)

def tampilkan_laporan(data_panen):
  total = hitung_total_panen(data_panen)
  rata_rata = hitung_rata_rata_panen(data_panen)
  print(f"Total hasil panen: {total} kg")
  print(f"Rata-rata hasil panen: {rata_rata:.2f} kg per petak")
def main():
  data_panen = input_data_panen()
  tampilkan_laporan(data_panen)
main()
```
